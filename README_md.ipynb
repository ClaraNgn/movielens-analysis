# MovieLens 100K Collaborative Filtering Analysis

Testing whether user-based or item-based KNN achieves lower prediction
error on the MovieLens 100K dataset.
theoreti

## Research Question

On the MovieLens 100K dataset, which collaborative filtering algorithm
achieves lower RMSE: user-based KNN or item-based KNN?
## Dataset

**MovieLens 100K** — 100,000 ratings from 943 users on 1,682 movies,
collected by GroupLens Research at University of Minnesota. Publicly
available at [grouplens.org/datasets/movielens](https://grouplens.org/datasets/movielens/).

After cleaning (removing movies with fewer than 5 ratings), the working
dataset contains **97,232 ratings across 943 users and 1,349 movies**.

## Methodology

1. **Data cleaning:** Filtered out 333 movies with fewer than 5 ratings
   (approximately 20% of catalog), retaining 97% of ratings.
2. **Exploratory analysis:** Distribution analysis of ratings, user
   activity, and movie popularity.
3. **Train/test split:** 80/20 split with random_state=42.
4. **Model training:** Two `KNNBasic` models from the `surprise` library,
   using cosine similarity — one user-based, one item-based.
5. **Evaluation:** RMSE on the test set.

## Key Findings

### 1. Long-tail user activity
Median user rated 65 movies; 25% of users rated ≤33 movies while the
most active user rated 737. This asymmetry follows a classic long-tail
distribution common in consumer behavior data.

![Ratings per User](chart2_ratings_per_user.png)

### 2. Sparse item distribution
25% of movies received ≤6 ratings, with some movies having only 1
rating before cleaning — highlighting the cold-start problem.

### 3. Positive rating bias
83% of ratings are 3 stars or above; rating 4 is most common (34%).
A naive baseline predicting mean rating (3.53) yields RMSE ≈ 1.12.

![Rating Distribution](chart1_rating_distribution.png)

### 4. User-based KNN outperforms item-based
| Model | RMSE |
|-------|------|
| User-based KNN (cosine) | **1.0040** |
| Item-based KNN (cosine) | 1.0204 |
| Naive baseline (predict mean) | ~1.12 |

Both algorithms significantly beat the baseline, but user-based
achieved 1.6% lower error than item-based.

## Discussion


1. **User-to-item structural difference:** MovieLens has fewer effective
   items than users, and user-based similarity may exploit denser
   neighborhoods.
2. **Selectivity operationalization:** MovieLens users rate movies
   post-consumption (evaluation), while Darani & Aghaie's model assumes
   selectivity at the point of choice (selection).
3. **Similarity metric under sparsity:** With ~7% matrix density, cosine
   similarity across users may be more stable than across items.

This suggests the theoretical proposition holds under specific data
regimes — a boundary condition worth investigating further.

## Limitations

- Only `KNNBasic` tested; mean-centering variants may change ranking
- Single random split; no cross-validation
- Cosine similarity only; Pearson or MSD may yield different results
- No user segmentation by selectivity to directly test the paper's claim
